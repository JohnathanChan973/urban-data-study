```python
from socrata_interface.domain import Domain
import socrata_interface.transformers as transform
from lazo.dataset_sketch_process import DatasetSketchProcessor
from lazo.lazo_service import LAZOService
```

```python
weho = Domain("data.weho.org")
lazo = LAZOService()
large_rows = ['phsm-35qu',
 '646y-yn7s',
 'e46y-98wp',
 'by9g-vc9g',
 'yai6-dpmb',
 'jaed-p39k',
 'yjid-9pyh',
 'b8ph-69yc',
 'cfmx-tb5w',
 'cz69-a684',
 '7kgt-b5t8',
 'vyfp-q9vs',
 'y2yz-icwv'] # Datasets with large amounts of rows
```

```python
for large_id in large_rows:
    large_meta = weho.metadata(large_id)
    large_schema = transform.extract_schema(large_meta)
    large_cols = transform.extract_joinable_columns(large_schema)
    large_gen = weho.dataset(large_id, True)
    dsp = DatasetSketchProcessor.process_generator(large_id, large_cols, large_gen)
    lazo.stage_dataset(dsp)
```
