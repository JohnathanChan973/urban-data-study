```python
weho = Domain("data.weho.org")
```

```python
weho_summary = DataSummary()
```

```python
ids = weho.dataset_ids()
```

```python
large_rows = ['phsm-35qu',
 '646y-yn7s',
 'e46y-98wp',
 'by9g-vc9g',
 'yai6-dpmb',
 'jaed-p39k',
 'yjid-9pyh',
 'b8ph-69yc',
 'cfmx-tb5w',
 'cz69-a684',
 '7kgt-b5t8',
 'vyfp-q9vs',
 'y2yz-icwv']
```

```python
large_id = large_rows[0]
large_meta = weho.metadata(large_id)
large_schema = transform.extract_schema(large_meta)
large_cols = transform.extract_joinable_columns(large_schema)
large_gen = weho.dataset(large_id, True)
```

```python
from lazo.dataset_sketch_process import DatasetSketchProcessor

dsp = DatasetSketchProcessor.process_generator(large_id, large_cols, large_gen)
```

```python
dsp
```

```python
weho.dataset_count()
```

```python
for id in ids:
    meta = weho.metadata(id)
    relevant = transform.extract_relevant_metadata(meta)
    
    if relevant.get("display") == "table":
        schema = transform.extract_schema(meta)
        row_counts = weho.row_count(id)
        null_counts = weho.null_count(id, schema)
        sparseness = transform.extract_sparseness(row_counts, null_counts)
        tabular_metadata = transform.aggregate_tabular_metadata(schema, row_counts, sparseness)
        relevant.update(tabular_metadata)

    weho_summary.ingest(relevant)
        
```

```python
weho_summary
```

```python
from socrata_interface.io import write_data

write_data(str(weho), "12345", "summary", weho_summary.to_dict())
```
